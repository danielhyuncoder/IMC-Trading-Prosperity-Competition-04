```python
# Worker functions
def get_clearing_price(bids: list[tuple[int]], asks: list[tuple[int]]):
    curr_best = 0
    c_price = 0
    
    for price_level in range(min(bids)[0], max(asks)[0]):
        bid_vol = 0
        ask_vol = 0
        for bid in bids:
            if bid[0]>=price_level:
                bid_vol+=bid[1]
        for ask in asks:
            if ask[0]<=price_level:
                ask_vol+=ask[1]
        
        traded_vol = min(bid_vol, ask_vol)
        if traded_vol >= curr_best:
            c_price=price_level
            curr_best=traded_vol
    return c_price, curr_best

def get_trade(bids: list[tuple[int]], asks: list[tuple[int]], clearing_price: int, sale_price: int, fee: float = 0.1):
    bid_vol = 0
    ask_vol = 0
    for bid in bids:
        if bid[0]>=clearing_price:
            bid_vol+=bid[1]
    for ask in asks:
        if ask[0]<=clearing_price:
            ask_vol+=ask[1]
    
    c_vol: int = 0
    best_profit: float = 0
    answer_price_level = 0
    answer_quantity = 0
    for bid in bids:
        if bid[0]>=clearing_price:
            c_vol += bid[1]
            volume_tradable = max(0, ask_vol - c_vol)
            profit = volume_tradable * (sale_price-bid[0])
            if profit >= best_profit:
                answer_price_level = bid[0]
                answer_quantity=volume_tradable
                best_profit=profit
    
    return answer_price_level, answer_quantity
```

```python
# DRYLAND FLAX DATA
flax_bids: list[tuple[int]] = [
        (30, 30000),
        (29, 5000),
        (28, 12000),
        (27, 28000)
]
flax_asks: list[tuple[int]] = [
        (28, 40000),
        (31, 20000),
        (32, 20000),
        (33, 30000),
]
sale_price_flax:int = 30
```

```python
# Clearing price for flax
get_clearing_price(flax_bids, flax_asks)
```

```python
# Get optimal trade for dry flax
for i in range(28, sale_price_flax):
    p=get_trade(flax_bids, flax_asks, i, sale_price_flax)
    print(str(p) + " | PNL: " + str((sale_price_flax-p[0]) * p[1]) )
```
